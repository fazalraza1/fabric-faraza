# 🥉 Bronze Layer — Ingest Raw Loan Transactions

**What we're doing:** Loading the raw CSV file exactly as-is. No changes, no filters.

This is our safety net — the original data is always preserved here.

```python
# Read the raw loan_transactions.csv file from the Lakehouse Files area into a Spark DataFrame
# Keep schema inference enabled so Bronze captures column types from the source file automatically
df_bronze = spark.read.option('header', 'true').option('inferSchema', 'true') \
    .csv('Files/loan_transactions.csv')

# Count all raw rows to confirm the Bronze ingest loaded every transaction from the CSV
print(f'✅ Bronze: Loaded {df_bronze.count()} raw transactions')
# Display a small sample of the unmodified Bronze records for a quick validation check
df_bronze.show(10)
```

```python
# Write the raw DataFrame to the Bronze Delta table without applying any cleansing rules
df_bronze.write.format('delta').mode('overwrite').saveAsTable('bronze_loan_transactions')
# Confirm that the Bronze Delta table is ready for downstream Silver processing
print('✅ Bronze table saved!')
```

```python
%%sql
-- Count Bronze records by Status to verify the raw ingest preserved each loan state
SELECT Status, COUNT(*) as Count FROM bronze_loan_transactions GROUP BY Status
```
