# 🥉 Step 1 — Ingest Raw Fraud Transactions (Bronze)

Load all transactions as-is. Some are fraudulent — we will let the AI find them.

```python
# Load the fraud transactions CSV from the Lakehouse Files area into a Spark DataFrame
df_raw = spark.read.option('header','true').option('inferSchema','true') \
    .csv('Files/fraud_transactions.csv')

# Count all transactions and flagged fraud cases to confirm the dataset loaded correctly
total = df_raw.count()
fraud_count = df_raw.filter('IsFraud = 1').count()

# Print a quick load summary and preview sample records before saving to Delta
print(f'✅ Loaded {total} transactions ({fraud_count} fraudulent)')
df_raw.show(10)
```

```python
# Write the raw fraud transactions to the Bronze Delta table for downstream processing
df_raw.write.format('delta').mode('overwrite').saveAsTable('bronze_fraud_transactions')

# Confirm the Bronze table was created successfully
print('✅ Bronze table saved!')
```
