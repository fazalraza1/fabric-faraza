# Notebook 03 — Fabric Semantic Model & Power BI on Databricks Data

**Role:** BI Developer / Business Analyst

**What this shows:** Business users build semantic models and Power BI reports directly on top of the Databricks-written Delta tables — with no knowledge of Spark or Databricks required.

> 🗣️ **Talking Point:** This is the 'better together' payoff for business users. Databricks handles the heavy engineering work. Fabric surfaces it as a business-friendly semantic layer. The BI team never needs to open a Databricks notebook.

```python
# Import PySpark functions for building Gold aggregation tables that feed the semantic model
from pyspark.sql import functions as F
print('Building Fabric semantic layer on top of Databricks Delta data...')
```

## Step 1 — Build Gold KPI Table for the Semantic Model

> 🗣️ **Talking Point:** Fabric's Direct Lake mode reads Gold Delta tables at near-real-time speed directly into Power BI — no import, no scheduled refresh, no stale dashboards. This is only possible because the underlying format is Delta.

```python
# Load the Fabric-enriched Gold transactions written in Notebook 02
gold_txn = spark.table('fabric_gold_transactions')

# Aggregate transaction KPIs by Region, Channel, and RiskTier for the semantic model
# These aggregates power the Power BI executive dashboard page
kpi_by_region = gold_txn \
    .groupBy('Region', 'Channel', 'RiskTier', 'AmountBand') \
    .agg(
        F.count('TransactionID').alias('NumTransactions'),
        F.round(F.sum('Amount'), 2).alias('TotalAmount'),
        F.round(F.avg('Amount'), 2).alias('AvgAmount'),
        F.sum(F.when(F.col('Status') == 'flagged', 1).otherwise(0)).alias('FlaggedCount'),
        F.sum(F.when(F.col('IsLargeTransaction') == True, F.col('Amount')).otherwise(0)).alias('LargeTransactionVolume')
    ) \
    .orderBy('Region', 'Channel')

# Write to a Gold KPI table that the Fabric Semantic Model will reference
kpi_by_region.write.format('delta').mode('overwrite').saveAsTable('gold_kpi_by_region')

# Preview the KPI aggregations
print(f'gold_kpi_by_region: {kpi_by_region.count()} rows')
kpi_by_region.show(10)
```

## Step 2 — Build Risk Summary Table

> 🗣️ **Talking Point:** Risk teams can query this table directly from Excel, Power BI, or the Fabric SQL endpoint — no Spark skills needed. Databricks handles the scale; Fabric democratises the access.

```python
# Build a risk summary aggregated by RiskTier and TransactionType for the risk management dashboard
risk_summary = gold_txn \
    .groupBy('RiskTier', 'TransactionType') \
    .agg(
        F.count('TransactionID').alias('TransactionCount'),
        F.round(F.sum('Amount'), 2).alias('TotalExposure'),
        F.round(F.avg('Amount'), 2).alias('AvgExposure'),
        F.countDistinct('AccountID').alias('UniqueAccounts'),
        F.countDistinct('CustomerID').alias('UniqueCustomers')
    ) \
    .orderBy('RiskTier', 'TransactionType')

# Write risk summary as a Delta table for the Fabric Semantic Model risk page
risk_summary.write.format('delta').mode('overwrite').saveAsTable('gold_risk_summary')

# Preview the risk summary output
print(f'gold_risk_summary: {risk_summary.count()} rows')
risk_summary.show()
```

```python
# List all Gold tables written by both pipelines to confirm semantic model inputs are ready
tables = ['db_silver_transactions', 'fabric_gold_transactions', 'gold_kpi_by_region', 'gold_risk_summary']

print('=== Tables ready for Fabric Semantic Model ===')
for t in tables:
    try:
        count = spark.table(t).count()
        print(f'  {t:40} {count:>6} rows  OK')
    except Exception as e:
        print(f'  {t:40} ERROR: {e}')

print()
print('Next step: Open Fabric Lakehouse -> New Semantic Model -> select gold_* tables')
print('Then open Power BI and connect via Direct Lake for real-time reporting.')
```
