# 🥉 Bronze Layer — Ingest Raw Loan Transactions

**What we're doing:** Loading the raw CSV file exactly as-is. No changes, no filters.

This is our safety net — the original data is always preserved here.

```python
# Load raw CSV — keep everything, even failed transactions
df_bronze = spark.read.option('header', 'true').option('inferSchema', 'true') \
    .csv('Files/loan_transactions.csv')

print(f'✅ Bronze: Loaded {df_bronze.count()} raw transactions')
df_bronze.show(10)
```

```python
# Save as Bronze Delta Table — never modify this table
df_bronze.write.format('delta').mode('overwrite').saveAsTable('bronze_loan_transactions')
print('✅ Bronze table saved!')
```

```python
%%sql
-- How many transactions by status?
SELECT Status, COUNT(*) as Count FROM bronze_loan_transactions GROUP BY Status
```
